```python
url = 'https://pytorch.tips/coffee'
fpath = 'coffee.jpg'
req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
with urllib.request.urlopen(req) as resp, open(fpath, 'wb') as f:
    f.write(resp.read())
```
