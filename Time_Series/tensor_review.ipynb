### Tensor Creations with PyTorch
#### 1d Tensor Creation

```python
import torch
```

```python
x= torch.tensor(data=[1,2,3])
print(x)
```

#### 2d Tensor Creation

```python
x= torch.tensor(data=[[1,2,3], [4,5,6]])
print(x)
```

#### Declare with tensor type
Allows various math functions

```python
x = torch.tensor(data=[1.3,.5,6.3, 4.9], dtype=torch.float32, requires_grad=True)
print(x)
```

#### Random Tensor

```python
x = torch.rand(2,3,4)
print(x)
```

#### Reproducibility - Setting the random seed

```python
torch.manual_seed(23)

# or 

import random
random.seed(23)
```

```python
# Include this snippet to set a constant random state
import torch
import random

torch.manual_seed(23)  # pick a number
random.seed(23) # pick a numnber 
```
