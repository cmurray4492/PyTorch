```python
# Graph data
def plot_data_1d(x, y_gt):
    """Plot 1D Data"""
    plt.scatter(x, y_gt, s=20, c="k")
    plt.xlabel("x0", fontsize=24)
    plt.ylabel("y", fontsize=24)
    plt.tick_params(axis="both", which="major", labelsize=16)
    plt.show()
```

```python
plot_data_1d(x, y_gt)
```
